```python
valor=float(input('digite o valor do medicamento: '))
idade=int(input('digite a sua idade: '))
if idade<=12:
  desconto=(15*valor)/100
  valor_final=valor-desconto
  print('o valor do desconto foi: ', desconto)
  print('o valor com o desconto e de: ',valor_final)

elif idade>=60 and idade<=75:
  desconto=(20*valor)/100
  valor_final=valor-desconto
  print('o valor do desconto foi: ', desconto)
  print('o valor com o desconto e de: ',valor_final)

elif idade>75:
  desconto=(25*valor)/100
  valor_final=valor-desconto
  print('o valor do desconto foi: ', desconto)
  print('o valor com o desconto e de: ',valor_final)
else:
  print('o medicamento não recebera desconto')
  print('o valor do medicamento sera de ', valor)
```
