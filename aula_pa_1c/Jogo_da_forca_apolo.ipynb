```python
import random
def escolher_palavra():
    listap = ["casa", "amor", "vida", "handball", "notebook", "etec"]
    palavra = random.choice(listap)

    return palavra
def mostrar_palavra(palavra, letras_correta):
    exibicao = ""
    for letra in palavra:
        if letra in letras_correta:
            exibicao += letra + " "
        else:
            exibicao += "_ "

    return exibicao
def tentar_letra(letra, palavra, letras_correta, letras_erradas):
    if letra in palavra:
        if letra not in letras_correta:
            letras_correta.append(letra)
            print(f"parabens essa letra esta correta: {letra}.")
        else:
            print("essa letra ja foi usada.")
    else:
        if letra not in letras_erradas:
            letras_erradas.append(letra)
            print(f"essa letra '{letra}' não esta na palavra.")
        else:
            print("errou")
def verificar_vitoria(palavra, letras_correta):
    for letra in palavra:
        if letra not in letras_correta:
            return False

    return True
def jogo_da_forca():
    palavra = escolher_palavra()
    letras_certas = []
    letras_erradas = []
    tentativas = 6

    print("JOGO DA FORCA")
    print("descubra a palavra")
    print(mostrar_palavra(palavra, letras_certas))
    print("*************************")

    while True:
        letra = input("Digite uma letra: ").lower()
        if len(letra) != 1 or not letra.isalpha():
            print("Digite apenas uma letra!")

            continue
        tentar_letra(letra, palavra, letras_certas, letras_erradas)
        print("\nPalavra:", mostrar_palavra(palavra, letras_certas))
        print("Letras erradas:", ", ".join(letras_erradas))
        print(f"Tentativas restantes: {tentativas - len(letras_erradas)}")

        print("***********************")
        if verificar_vitoria(palavra, letras_certas):
            print("vc venceuu")
            print(f"palavra: {palavra}")

            break
        if len(letras_erradas) >= tentativas:
            print("Você perdeu")
            print(f"palavra: {palavra}")

            break

jogo_da_forca()
```
