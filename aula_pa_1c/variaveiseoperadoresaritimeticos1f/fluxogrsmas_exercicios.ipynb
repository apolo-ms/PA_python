```python
#simulação de investimento
valorinvestido=float(input("digite o valor investido: "))
taxano=float(input("digite valor taxa: "))
tempo=int(input("digite o ano: "))
valorfinal=valorinvestido*(1+taxano)**tempo
print("o valor final e: ",valorfinal)
```
