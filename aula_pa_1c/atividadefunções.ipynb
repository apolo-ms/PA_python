6.calculo de desconto

```python
def calcular_desconto(valor_original,percentual_de_desconto):
  porcentagem=percentual_de_desconto/100
  valor_a_ser_descontado=valor_original * porcentagem
  valor_final=valor_original-valor_a_ser_descontado
  return valor_final

percentual_de_desconto=int(input("digite o percentual de desconto sem o sibolo de porcentagem "))
valor_original=float(input("digitr um valor "))
# resultado=calcular_desconto()
print(f"o valor final e de {calcular_desconto(valor_original,percentual_de_desconto)}R$")
```

8.calculadora simples

```python
def calculadora(a, b, operacao):
    if operacao == "soma":
        return a + b
    elif operacao == "subtração":
        return a - b
    elif operacao == "divisão":
        return a / b
    elif operacao == "multiplicação":
        return a * b
    else:
        return None
operacoes_possiveis = ["soma", "subtração", "divisão", "multiplicação"]
operacao = input("Escolha uma operação entre soma, subtração, divisão e multiplicação: ")

if operacao not in operacoes_possiveis:
    print("A operação não pode ser executada.")
else:
    a = float(input("Digite o primeiro número: "))
    b = float(input("Digite o segundo número: "))

    if operacao == "divisão" and b == 0:
        print("A conta não pode ser executada (divisão por zero).")
    else:
        resultado = calculadora(a, b, operacao)
        print(f"O resultado da {operacao} foi {resultado}")
```

9.mensagem automatica


```python
def enviar_email(destinatario, assunto, corpo="Este é um email automático."):
    print("----- Enviando Email -----")
    print(f"Destinatário: {destinatario}")
    print(f"Assunto: {assunto}")
    print(f"Corpo: {corpo}")
    print("--------------------------")


# Programa principal
destinatario = input("Digite o email do destinatário: ")
assunto = input("Digite o assunto do email: ")

# Pergunta se o usuário quer escrever um corpo personalizado
escolha = input("Deseja escrever o corpo do email? s/n: ")

if escolha.lower() == "s":
    corpo = input("Digite o corpo do email: ")
    enviar_email(destinatario, assunto, corpo)
else:
    enviar_email(destinatario, assunto)
```

10.sistema de notas


```python
def avaliar_aluno(nota):
    if nota >= 7:
        return "Aprovado"
    elif nota >= 5 and nota < 7:
        return "Recuperação"
    else:
        return "Reprovado"
nota = float(input("Digite a nota do aluno: "))
resultado = avaliar_aluno(nota)
print("Situação do aluno:", resultado)
```
